# Latihan 2 Praktikum Persamaan Diferensial Biasa: Metode Taylor dan Metode Runge-Kutta

## Soal Kelas A

Diberikan suatu masalah nilai awal (*Initial Value Problem*):

$$
\begin{align*}
    y' &= e^{2t} + 2y,\\
       & y(0) = 2 \nonumber
\end{align*}
$$

Diberikan pada interval $0\leq t\leq 2$ dengan stepsize $h = 0.1$

Diketahui bahwa solusi eksak dari model diberikan oleh

$$
\begin{align*}
    y(t) = (t+2)e^{2t}
\end{align*}
$$

## Soal Kelas B

Diberikan suatu masalah nilai awal (*Initial Value Problem*):

$$
\begin{align*}
    y' &= e^{t-y},\\
       & y(0) = 1 \nonumber
\end{align*}
$$

Diberikan pada interval $0\leq t\leq 3$ dengan stepsize $h = 0.2$

Diketahui bahwa solusi eksak dari model diberikan oleh

$$
\begin{align*}
    y(t) = \ln(e^t + e - 1)
\end{align*}
$$
