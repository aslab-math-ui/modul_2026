# Latihan 3 Praktikum Persamaan Diferensial Biasa: Metode Multistep

## Soal

Diberikan suatu masalah nilai awal (*Initial Value Problem*):

$$
\begin{align*}
    y' &= \frac{y^{2}+y}{t},\\
       & y(1) = -2 \nonumber
\end{align*}
$$

Diberikan pada interval $1\leq t\leq 3$ dengan stepsize $h = 0.1$

Diketahui bahwa solusi eksak dari model diberikan oleh

$$
\begin{align*}
    y(t) = \frac{2t}{1-2t}
\end{align*}
$$

- Untuk Kelas A, gunakan 2 metode berikut :
    1. Adams-Bashforth orde-2 (Two-step) dengan perhitungan nilai awal menggunakan metode **Runge-kutta orde 4** 
    2. Adams Predictor-Corrector Orde-4 dengan perhitungan nilai awal menggunakan metode **euler**.
    
- Untuk Kelas B, gunakan 2 metode berikut :
    1. Adams-Bashforth orde-3 (Three-step) dengan perhitungan nilai awal menggunakan metode **euler**.
    2. Adams Predictor-Corrector Orde-3 dengan perhitungan nilai awal menggunakan metode **Runge-kutta orde 4**.
