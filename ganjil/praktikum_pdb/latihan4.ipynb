# Latihan 4 Praktikum Persamaan Diferensial Biasa: Sistem PDB dan Orde Tinggi

## Soal

Diberikan suatu masalah nilai awal (*Initial Value Problem*):

$$
\begin{align*}
   & y''' + y'' - 4y' - 4y = 0,\\
      & y(0) = 3 \nonumber, \\
      & y'(0) = -1 \nonumber, \\
      & y''(0) = 9 \nonumber, \\
\end{align*}
$$

Diberikan pada interval $0\leq t\leq 2$ dengan stepsize $h = 0.1$

Diketahui bahwa solusi eksak dari model diberikan oleh

$$
\begin{align*}
    y(t) = e^{-t} + e^{2t} + e^{-2t}
\end{align*}
$$

Gunakanlah metode Runge-Kutta 4 untuk mengaproksimasi solusi dari y, y', dan y''. Bandingkan dengan solusi eksak menggunakan table dan plot
