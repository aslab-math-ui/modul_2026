Dalam coding, kita sering kali perlu mengukur seberapa efisien kode yang kita buat. Dua konsep utama yang digunakan untuk mengukur efisiensi ini adalah *Runtime* dan *Time Complexity*.

1. **Runtime**
   - Merupakan waktu nyata (dalam detik, milidetik, atau mikrodetik) yang dibutuhkan komputer untuk menjalankan suatu program.
   - Nilai *runtime* bersifat absolut dan sangat bergantung pada faktor eksternal seperti spesifikasi perangkat keras (CPU, RAM), bahasa pemrograman yang digunakan, compiler/interpreter, serta beban proses lain yang sedang berjalan di sistem pada saat yang bersamaan.
   - Oleh karena itu, mengukur efisiensi hanya dengan *runtime* kurang objektif untuk membandingkan performa algoritma secara teoretis.

2. **Time Complexity**
   - Merupakan representasi matematika yang menggambarkan bagaimana kebutuhan waktu sebuah algoritma tumbuh seiring dengan bertambahnya ukuran data masukan ($N$).
   - Kompleksitas waktu diukur menggunakan notasi asimtotik seperti **Big-O** (misalnya $O(1)$, $O(n)$, $O(n^2)$).
   - Berbeda dengan *runtime*, kompleksitas waktu bersifat **independen** terhadap spesifikasi mesin komputer. Ini memberikan standar objektif untuk menganalisis batas kinerja efisiensi suatu algoritma pada skenario data skala besar.

Pada modul ini, kita akan mempelajari cara mengukur *runtime* secara presisi menggunakan modul bawaan `timeit`, serta bagaimana melakukan estimasi *time complexity* secara empiris menggunakan modul `big-O`.

## Modul *timeit*


Modul `timeit` adalah modul *built-in* di Python yang dirancang khusus untuk mengukur waktu eksekusi dari potongan kode kecil. Modul ini sangat berguna untuk melakukan *benchmarking* atau membandingkan *runtime* dari beberapa *snippet* kode atau algoritma.

Untuk menggunakan `timeit`, cukup import saja

````python
import timeit
````

### Contoh penggunaan `timeit`

Berikut adalah contoh penggunaan dari `timeit`

````python
def hitung():
    total = 0
    for i in range(100):
        total += i
    return total

execution_time = timeit.timeit(hitung)

print("Total time:", execution_time)
````

Kode di atas menghitung total runtime yang dijalankan oleh fungsi `hitung`, tapi kenapa bisa runtimenya bisa melebihi 2 detik padahal seharusnya komputer bisa menghitung dengan sangat cepat? Hal itu terjadi karena fungsi `timeit.timeit()` nge-run kode tersebut sebanyak 1.000.000 kali secara default. Apabila kita ingin mengganti angka tersebut, maka kita dapat mengubah parameter `number`.

Tentu fungsi `timeit.timeit()` memiliki banyak parameter lainnya. Syntax lengkapnya dapat dilihat sebagai berikut:

```python
timeit.timeit(stmt='pass', setup='pass', timer=<default timer>, number=1000000, globals=None)
```

Berikut adalah penjelasan dari setiap parameter

1. **`stmt` (Statement)**:
   - Kode yang ingin diukur waktu eksekusinya.
   - Bisa berupa **string** berisi kode Python, atau langsung merujuk ke nama objek **fungsi** (*callable*) seperti contoh di atas (`hitung` tanpa tanda kurung).

2. **`setup`**:
   - Kode persiapan yang dijalankan sekali sebelum pengukuran waktu dimulai (tidak ikut dihitung waktunya).
   - Biasanya digunakan untuk mengimpor modul atau mendefinisikan variabel awal yang dibutuhkan oleh `stmt`.
   - Contoh:
     ```python
     timeit.timeit(stmt='math.sqrt(100)', setup='import math')
     ```

3. **`number`**:
   - Jumlah iterasi (berapa kali kode di `stmt` dijalankan).
   - Secara *default* bernilai `1.000.000` kali.

4. **`globals`**:
   - Menentukan ruang nama (*namespace*) tempat kode dijalankan.
   - Jika ingin menggunakan fungsi atau variabel yang sudah didefinisikan di dalam file/notebook tanpa menulis `setup`, cukup menggunakan `globals=globals()`.


 Output/Return Value:
- Fungsi ini mengembalikan nilai berupa **float** yang menunjukkan total akumulasi waktu (dalam satuan detik) yang dibutuhkan untuk menjalankan pernyataan (`stmt`) sebanyak `number` kali.


Sekarang, kita akan melihat contoh penggunaan `timeit` untuk mengukur runtime dari salah satu algoritma yang dipelajari pada pertemuan pertama, yaitu linear search. Ingat kembali linear search memiliki algoritma sebagai berikut.

````python
def linear_search(arr, key):
    for i in range(0, len(arr)):
        if arr[i] == key:
            return i

    # sampai sini, berarti elemen tidak ditemukan
    return -1
````

````python
A = [1, 5, 2, 3, 4, 8, 7, 6, 10, 9]
linear_search(A, 8)
````

Misal kita ingin menghitung rata-rata waktu eksekusi linear search tersebut

````python
execution_time = timeit.timeit(
    "linear_search(A, 8)",    # Statement yang ingin diukur runtimenya
    number=10000,             # Statement dirun sebanyak 10 000 kali
    globals=globals()         # Karena statement menggunakan variabel dari luar statement, yaitu 'A',
)                             # maka perlu ditulis line ini

# Notes: output berikut berupa dalam satuan detik
print("Total time:", execution_time)
print("Average time:", execution_time/10000)
````

Perhatikan bahwa karena kita atur parameter `number` agar bernilai 10 000. Oleh karena itu, kita membagi variabel `execution_time` dengan 10 000 untuk memperoleh rata-ratanya.

### Magic Commands

Magic commands untuk `timeit` merupakan perintah spesial yang ada pada JupyterNnotebook, JupyterLab, dan Google Colab. Terdapat 2 jenis magic command yang ada untuk `timeit`, yaitu:

- `%timeit` mengukur waktu eksekusi pada 1 baris kode saja.

- `%%timeit` mengukur waktu eksekusi pada 1 cell.

Berikut adalah contoh penggunaannya

````python
# timeit TIDAK menghitung runtime dari instantiasi variabel A
A = [1, 5, 2, 3, 4, 8, 7, 6, 10, 9]
%timeit linear_search(A, 8)
````

````python
# timeit MENGHITUNG runtime dari instantiasi variabel A
%%timeit
A = [1, 5, 2, 3, 4, 8, 7, 6, 10, 9]
linear_search(A, 8)
````

Perhatikan bahwa `%%timeit` menghitung runtime pada keseluruhan cellnya hingga bisa saja runtime yang dihitung tidak akurat karena ada kode lain yang berjalan juga. Pada kasus ini, kode `A = [1, 5, 2, 3, 4, 8, 7, 6, 10, 9]` akan menambah pada runtime.

## Modul *big-O*

Jika `timeit` digunakan untuk mengukur waktu eksekusi (*runtime*) secara absolut dalam satuan detik, modul **`big-O`** adalah library Python yang dirancang untuk memperkirakan kompleksitas waktu asimtotik (seperti $O(1)$, $O(\log n)$, $O(n)$, $O(n \log n)$, $O(n^2)$) dari sebuah fungsi.

Modul ini bekerja dengan cara:
1. Menjalankan fungsi target dengan berbagai ukuran input ($N$).
2. Mengukur waktu eksekusi untuk setiap ukuran input tersebut.
3. Melakukan *curve fitting* (pencocokan kurva) terhadap model-model matematika kompleksitas waktu populer.
4. Menentukan model mana yang paling cocok (paling akurat menggambarkan pertumbuhan waktu terhadap ukuran input).

### Contoh Penggunaan `big-O`

Untuk mengukur kompleksitas waktu dari sebuah fungsi, kita akan menggunakan `bigo.BigO`. Kita juga perlu mendefinisikan generator data masukan agar modul `big-O` dapat membuat input acak dengan ukuran $N$ yang bervariasi secara otomatis.

Berikut adalah contoh penggunaannya untuk menguji kompleksitas waktu.

````python
# Menghitung sumasi dari semua elemen pada array
def total(arr):
  sum = 0
  for x in arr:
    sum += x
  return sum

# Kita definisikan fungsi untuk generate data yang berukuran n
def data_generator(n):
  return big_o.datagen.integers(n, min_=0, max_=1000)

# Mengukur kompleksitas waktu menggunakan generator bawaan dari big-O
best, others = big_o.big_o(
    total,
    data_generator,
)

print(f"Kompleksitas waktu terbaik adalah {best}")
````

Didapat hasil curve fitting terbaik untuk `total` adalah  $O(n)$

### Penjelasan Hasil dan Parameter Utama

Dalam fungsi `big_o.big_o()`, kita menggunakan beberapa komponen penting:

#### Parameter Wajib:
1. **Fungsi Target**:
   - Fungsi yang ingin kita ukur kompleksitas waktunya. Fungsi ini harus menerima satu parameter berupa data masukan yang dihasilkan oleh generator.

2. **Generator Data**:
   - Fungsi pembantu yang menerima ukuran masukan `n` dan mengembalikan data (misalnya array bilangan acak) dengan ukuran `n` tersebut.
   - Kita menggunakan generator bawaan `big_o.datagen.integers` untuk membuat array berisi bilangan bulat acak.

#### Parameter Opsional:

* **`min_n`**:
  - Ukuran input minimum yang digunakan untuk pengujian (nilai *default*-nya adalah `10`).

* **`max_n`**:
  - Ukuran input maksimum yang digunakan untuk pengujian (nilai *default*-nya adalah `100000`).

* **`n_measures`**:
  - Jumlah langkah/titik ukuran input dari `min_n` hingga `max_n` yang akan diuji (nilai *default*-nya adalah `10`).

* **`n_timings`**:
  - Jumlah pengulangan eksekusi fungsi untuk setiap ukuran input $N$ untuk kemudian diambil waktu terkecilnya guna mengurangi bias *noise* dari sistem (nilai *default*-nya adalah `5`).

`big_o.big_o()` masih memiliki beberapa parameter opsional lainnya yang tidak dibahas pada modul ini. Daftar lengkap semua parameter dapat dilihat dalam dokumentasi `big-O`.

#### Output

Fungsi `big_o.big_o()` memberikan dua output utama:

* **`best` (Hasil Terbaik)**:
  - Berisi kelas kompleksitas waktu (seperti `Linear`, `Constant`, `Quadratic`) yang paling cocok (*best-fitting*) dengan data eksperimen.
  - Nilai ini menyertakan hasil pencocokan kurva (*curve fitting*), misalnya: `Linear: time = 0.00015 + 3.4E-08*n (sec)`.

* **`others` (Alternatif Kompleksitas Lain)**:
  - Berisi sebuah *dictionary* (kamus) dari kelas-kelas kompleksitas lainnya beserta nilai R-squared (atau *residual sum of squares*)-nya.
  - Nilai ini menunjukkan seberapa dekat setiap model matematika alternatif dengan data waktu riil. Semakin kecil nilai *residual error* atau galatnya, semakin cocok kelas kompleksitas tersebut dengan perilaku fungsi kita.
  - Namun, biasanya kita hanya akan melihat yang best saja.

Berikut adalah tabel jenis-jenis kelas kompleksitas waktu (*Time Complexity*) utama yang diuji oleh modul `big-O`:

| Nama | Notasi Big-O | Objek `big_o` |
| :--- | :--- | :--- |
| Constant | $O(1)$ | `big_o.complexities.Constant` |
| Logarithmic | $O(\log n)$ | `big_o.complexities.Logarithmic` |
| Linear | $O(n)$ | `big_o.complexities.Linear` |
| Linearithmic | $O(n \log n)$ | `big_o.complexities.Linearithmic` |
| Quadratic | $O(n^2)$ | `big_o.complexities.Quadratic` |
| Cubic | $O(n^3)$ | `big_o.complexities.Cubic` |
| Polynomial | $O(n^a)$ | `big_o.complexities.Polynomial` |

### Generator Data (`big_o.datagen`)

Modul `big-O` membutuhkan generator data untuk membuat *input* yang uji secara dinamis dengan ukuran $N$ yang bervariasi. Dua jenis generator yang paling sering kita gunakan adalah:

1. **`big_o.datagen.integers(n, min_, max_)`**:
   - **Kegunaan**: Digunakan untuk menguji fungsi yang menerima masukan berupa **list/array** berisi bilangan bulat.
   - **Cara Kerja**: Generator ini akan membuat sebuah list acak sepanjang `n` elemen. Setiap elemen di dalamnya berupa bilangan bulat acak bernilai antara `min_` sampai `max_`.

2. **`big_o.datagen.n_(n)`**:
   - **Kegunaan**: Digunakan untuk menguji fungsi yang parameter masukannya berupa **nilai integer tunggal** $N$ itu sendiri.
   - **Cara Kerja**: Generator ini hanya akan mengembalikan nilai bilangan bulat `n` itu sendiri secara langsung untuk diteruskan ke fungsi target.

**Notes:** untuk menggunakan data generator, harus menggunakan sebuah *wrapper function*. Pada contoh di atas, *wrapper function* tersebut adalah `data_generator` yang menerima 1 input berupa ukuran data (`n`) dan outputnya adalah jenis datagen yang digunakan.

Sekarang, kita akan menggunakan `big-O` untuk mencari kompleksitas waktu dari algoritma linear search sebelumnya.

````python
# Kita definisikan fungsi perantara agar sesuai dengan format yang dibutuhkan big-O
def linear_search_test(arr):
    key = arr[-1]   # Ambil elemen terakhir untuk menguji skenario worst-case
    return linear_search(arr, key)

# Kita definisikan fungsi untuk generate data yang berukuran n
def data_generator(n):
    return big_o.datagen.integers(n, min_=0, max_=100000)

# Mengukur kompleksitas waktu menggunakan generator bawaan dari big-O
best, others = big_o.big_o(
    linear_search_test,
    data_generator,
    min_n=10,
    max_n=100000,
    n_measures=10,
    n_timings=5
)

print(f"Kompleksitas waktu terbaik adalah {best}")
````

Dapat dilihat hasil *curve fitting* terbaik untuk linear search adalah $O(n)$

Perhatikan bahwa fungsi yang diuji bukanlah `linear_search`, melainkan `linear_search_test` yang didefinisikan di atas. Hal ini perlu dilakukan karena `big_o.big_o()` hanya akan menguji fungsi yang memiliki 1 parameter, tapi fungsi `linear_search` memiliki 2 parameter, yaitu `arr` dan `key`. Untuk itu, dibuatlah sebuah *wrapper function* `linear_search_test` yang hanya menerima 1 input, yaitu `arr`.
