# No 2 - Stack

Sebuah robot menerima sekumpulan instruksi dalam bentuk list/array yang berisi perintah `"move"` dan `"undo"`.
- Perintah `"move"` berarti robot bergerak maju satu langkah.
- Perintah `"undo"` berarti robot membatalkan satu langkah gerakan `"move"` terakhir yang telah dilakukan.

Tugas Anda adalah membuat fungsi `is_valid_commands(commands)` dengan input berupa array (bukan yang numpy/sympy) menggunakan struktur data **Stack** untuk menentukan apakah urutan perintah yang diberikan valid.

**Hasil output:**
1. Perintah dianggap **tidak valid** (`False`) jika terdapat perintah `"undo"` ketika tidak ada langkah `"move"` yang bisa dibatalkan.
2. Jika seluruh urutan perintah berhasil diproses tanpa melanggar aturan di atas, maka kembalikan nilai `True`.

#### Contoh:
- Input: `["move", "move", "undo", "move"]` -> Output: `True`
- Input: `["undo", "move"]` -> Output: `False` (karena `undo` dilakukan di awal saat belum ada gerakan)
- Input: `["move", "undo", "undo"]` -> Output: `False` (karena `undo` kedua tidak memiliki pasangan `move` yang tersisa)

Setelah membuat fungsi tersebut, **jalankan kode berikut** untuk menguji beberapa contoh perintah yang valid dan tidak valid.

```python
commands_1 = ["move", "move", "move", "undo", "move", "undo", "move", "move", "undo", "move"]
commands_2 = ["move", "move", "undo", "undo", "undo", "move", "move", "move", "undo", "move"]
commands_3 = ["move", "move", "undo", "move", "move", "undo", "undo", "move", "move", "undo", "undo"]
commands_4 = ["undo", "move", "move", "move", "undo", "move", "move", "undo", "undo", "move"]

print(f"Hasil perintah 1: {is_valid_commands(commands_1)}")
print(f"Hasil perintah 2: {is_valid_commands(commands_2)}")
print(f"Hasil perintah 3: {is_valid_commands(commands_3)}")
print(f"Hasil perintah 4: {is_valid_commands(commands_4)}")
```
