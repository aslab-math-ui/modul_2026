## Soal 1 — Diferensiasi Numerik & Ekstrapolasi Richardson

### 1a. Fungsi `turunkan(expr, x0, h)` &nbsp; *(20 poin)*

Buatlah fungsi `turunkan(expr, x0, h)` yang menerima:

| Parameter | Tipe | Keterangan |
|-----------|------|------------|
| `expr` | ekspresi SymPy | fungsi $f(x)$ secara simbolik |
| `x0` | `float` | titik evaluasi |
| `h` | `float` | stepsize |

Fungsi menghitung dan menampilkan **tabel** berisi:
- Nilai **eksak** $f'(x_0)$ menggunakan `sympy.diff`
- Aproksimasi dari **5 metode** berikut, beserta error absolutnya:

<div class="formula-box">

$$
\def\arraystretch{2.5}
\begin{array}{|c|l|c|c|}
\hline
\textbf{No} & \textbf{Metode} & \textbf{Rumus} & \textbf{Order} \\
\hline
1 & \text{Forward/Backward Difference} & \dfrac{f(x_0+h)-f(x_0)}{h} & O(h) \\
2 & \text{Three-Point Endpoint} & \dfrac{-3f(x_0)+4f(x_0+h)-f(x_0+2h)}{2h} & O(h^2) \\
3 & \text{Three-Point Midpoint} & \dfrac{f(x_0+h)-f(x_0-h)}{2h} & O(h^2) \\
4 & \text{Five-Point Endpoint} & \dfrac{-25f(x_0)+48f(x_0+h)-36f(x_0+2h)+16f(x_0+3h)-3f(x_0+4h)}{12h} & O(h^4) \\
5 & \text{Five-Point Midpoint} & \dfrac{f(x_0-2h)-8f(x_0-h)+8f(x_0+h)-f(x_0+2h)}{12h} & O(h^4) \\
\hline
\end{array}
$$

</div>

**Pengujian:**

Uji fungsi dengan **dua fungsi berbeda**:

1. **(Wajib)** $f(x) = \sin(\pi x)\cdot e^{-x/2}$, $x_0 = 0.5$, $h = 0.1$
2. **(Bebas)** satu fungsi lain pilihan sendiri

### 1b. Fungsi `ekstrapolasi(expr, x0, h_init)` &nbsp; *(20 poin)*

Buatlah fungsi `ekstrapolasi(expr, x0, h_init)` yang menerapkan **Ekstrapolasi Richardson** pada metode **Three-Point Midpoint**, menghasilkan tabel dari $O(h^2)$ hingga $O(h^8)$.

**Rumus rekursi Richardson** (dari $O(h^{2j})$ ke $O(h^{2(j+1)})$):

$$D_{i,j} = \frac{4^j \cdot D_{i,j-1} - D_{i-1,j-1}}{4^j - 1}$$

di mana $D_{i,0}$ adalah hasil Three-Point Midpoint dengan $h = h_{\text{init}} / 2^i$.

**Format tabel yang diharapkan** (4 baris × 4 kolom, lower triangular):

<div class="formula-box">

| $h$ | $O(h^2)$ | $O(h^4)$ | $O(h^6)$ | $O(h^8)$ |
|-----|----------|----------|----------|----------|
| $h_\text{init}$ | $D_{0,0}$ | | | |
| $h_\text{init}/2$ | $D_{1,0}$ | $D_{1,1}$ | | |
| $h_\text{init}/4$ | $D_{2,0}$ | $D_{2,1}$ | $D_{2,2}$ | |
| $h_\text{init}/8$ | $D_{3,0}$ | $D_{3,1}$ | $D_{3,2}$ | $D_{3,3}$ |

</div>

**Uji** dengan $f(x) = \sin(\pi x)\cdot e^{-x/2}$, $x_0 = 0.5$, $h_{\text{init}} = 0.4$.

## Soal 2 — Integrasi Numerik

Hitunglah nilai integral berikut:

$$\int_0^2 x\, e^{-x^2}\, dx$$

Nilai eksak dapat diperoleh secara analitik menggunakan substitusi $u = x^2$:

$$\int_0^2 x\, e^{-x^2}\, dx = \frac{1 - e^{-4}}{2}$$

### 2a. Closed Newton-Cotes &nbsp; *(15 poin)*

Gunakan keempat metode **closed Newton-Cotes** berikut **secara langsung pada interval $[0,2]$** (tanpa membagi subinterval):

<div class="formula-box">

$$
\def\arraystretch{2.5}
\begin{array}{|c|l|c|c|}
\hline
\textbf{No} & \textbf{Metode} & \textbf{Rumus} & \textbf{Order Error} \\
\hline
1 & \text{Trapezoidal} & \dfrac{h}{2}[f(a)+f(b)] & O(h^2) \\
2 & \text{Simpson 1/3} & \dfrac{h}{6}[f(a)+4f(m)+f(b)],\ m=\dfrac{a+b}{2} & O(h^4) \\
3 & \text{Simpson 3/8} & \dfrac{3h}{8}[f(a)+3f(a+h)+3f(a+2h)+f(b)],\ h=\dfrac{b-a}{3} & O(h^4) \\
4 & \text{Boole's Rule} & \dfrac{2h}{45}[7f_0+32f_1+12f_2+32f_3+7f_4],\ h=\dfrac{b-a}{4} & O(h^6) \\
\hline
\end{array}
$$

</div>

Tampilkan **tabel hasil** beserta error absolut dari masing-masing metode.

### 2b. Composite Simpson $\tfrac{1}{3}$ &nbsp; *(15 poin)*

Gunakan **Composite Simpson $\tfrac{1}{3}$** dengan $n = 2, 4, 8, 16, 32$ subinterval.

**Rumus:**

$$\int_a^b f(x)\,dx \approx \frac{h}{3}\left[f(x_0) + 4\!\sum_{\text{ganjil}} f(x_i) + 2\!\sum_{\text{genap}} f(x_i) + f(x_n)\right], \qquad h = \frac{b-a}{n}$$

<div class="callout warn">
  <span class="callout-icon">⚠️</span>
  <div>$n$ harus bilangan <strong>genap</strong> untuk Composite Simpson $\tfrac{1}{3}$.</div>
</div>

## Soal 3 — Sistem Persamaan Linear

Diberikan SPL $A\mathbf{x} = \mathbf{b}$ dengan $n = 4$ persamaan. Terlebih dahulu, hitung nilai $k$ dari NPM:

$$k = \text{NPM} \bmod 3$$

Kemudian konstruksi matriks $A$ dan vektor $\mathbf{b}$ sebagai berikut ($i, j = 1,2,3,4$):

<div class="formula-box">

$$A_{ij} = \begin{cases} (k+3)(2i-1) & \text{jika } i = j \\ \dfrac{1}{i+j+k-1} & \text{jika } i \neq j \end{cases}$$

$$b_i = \sum_{j=1}^{4} A_{ij}, \qquad i = 1,2,3,4$$

</div>

<div class="callout tip">
  <span class="callout-icon">✅</span>
  <div><strong>Catatan penting:</strong> Solusi eksak dari SPL ini adalah $\mathbf{x} = [1,1,1,1]^T$, karena $b_i = \sum_j A_{ij}$ berarti $A\mathbf{1} = \mathbf{b}$. Gunakan ini untuk memverifikasi hasil Anda.</div>
</div>

### 3a. Metode Gauss-Seidel &nbsp; *(15 poin)*

Selesaikan SPL menggunakan **Metode Iteratif Gauss-Seidel** dengan parameter:

| Parameter | Nilai parameter penting|
|-----------|-------|
| Tebakan awal | $\mathbf{x}^{(0)} = [0,0,0,0]^T$ |
| Toleransi | $10^{-6}$ |
| Iterasi maksimum | 100 |

**Rumus pembaruan Gauss-Seidel:**

$$x_i^{(k)} = \frac{1}{a_{ii}}\left(b_i - \sum_{j<i} a_{ij}\, x_j^{(k)} - \sum_{j>i} a_{ij}\, x_j^{(k-1)}\right)$$

Tampilkan **tabel log iterasi** berisi nilai $\mathbf{x}^{(k)}$ di setiap iterasi beserta $\|\mathbf{x}^{(k)} - \mathbf{x}^{(k-1)}\|_\infty$.

### 3b. Faktorisasi LU — Metode Doolittle &nbsp; *(15 poin)*

Selesaikan SPL menggunakan **Faktorisasi LU Doolittle**, di mana $A = LU$ dengan:
- $L$: lower triangular dengan **diagonal = 1**
- $U$: upper triangular

**Algoritma** (untuk $i = 0, 1, \ldots, n-1$):

<div class="formula-box">

$$U_{ij} = A_{ij} - \sum_{k=0}^{i-1} L_{ik}U_{kj}, \qquad j = i,\, i{+}1,\ldots,\, n{-}1$$

$$L_{ji} = \frac{A_{ji} - \displaystyle\sum_{k=0}^{i-1} L_{jk}U_{ki}}{U_{ii}}, \qquad j = i{+}1,\ldots,\, n{-}1$$

</div>

Setelah mendapatkan $L$ dan $U$:
1. Selesaikan $L\mathbf{y} = \mathbf{b}$ dengan **forward substitution**
2. Selesaikan $U\mathbf{x} = \mathbf{y}$ dengan **backward substitution**

Tampilkan matriks $L$, $U$, dan solusi $\mathbf{x}$.

## Bonus — Faktorisasi LU Crout &nbsp; *(+10 poin)*

<div class="callout warn">
  <span class="callout-icon">✨</span>
  <div>Bagian ini <strong>tidak wajib</strong>. Kerjakan hanya setelah semua soal wajib diselesaikan.</div>

Selesaikan SPL yang sama menggunakan **Faktorisasi LU Crout**, di mana $A = LU$ dengan:
- $L$: lower triangular
- $U$: upper triangular dengan **diagonal = 1** *(kebalikan dari Doolittle)*

**Algoritma** (untuk $j = 0, 1, \ldots, n-1$):

<div class="formula-box">

$$L_{ij} = A_{ij} - \sum_{k=0}^{j-1} L_{ik}U_{kj}, \qquad i = j,\, j{+}1,\ldots,\, n{-}1$$

$$U_{ji} = \frac{A_{ji} - \displaystyle\sum_{k=0}^{j-1} L_{jk}U_{ki}}{L_{jj}}, \qquad i = j{+}1,\ldots,\, n{-}1$$

</div>

<div class="callout tip">
  <span class="callout-icon">💡</span>
  <div>Struktur algoritma sangat mirip dengan Doolittle — bedanya diagonal satuan ada di $U$ (bukan $L$). Fungsi <code>forward_sub</code> dan <code>backward_sub</code> yang sudah dibuat di 3b <strong>dapat digunakan kembali</strong>.</div>
</div>

Tampilkan matriks $L$, $U$, solusi $\mathbf{x}$, dan bandingkan hasil Crout dengan Doolittle.
